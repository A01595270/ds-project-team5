Variables categóricas

```python
vars_categoricas = df.select_dtypes(include='object').columns.tolist()
tabla_categoricas = pd.DataFrame({"Categóricas": vars_categoricas})

display(tabla_categoricas)
```

Frecuencias

```python
def tabla_categoricas(df, columnas):
    """Genera un resumen de variables categóricas."""

    resultados = []

    for columna in columnas:

        frecuencias = df[columna].value_counts()

        resultado = {
            "variable": columna,
            "n_categorias": df[columna].nunique(),
            "n_faltantes": df[columna].isnull().sum(),
            "moda": frecuencias.index[0],
            "top_2": frecuencias.index[1] if len(frecuencias) > 1 else "-",
            "top_3": frecuencias.index[2] if len(frecuencias) > 2 else "-"
        }

        resultados.append(resultado)

    return pd.DataFrame(resultados)

tabla_categoricas(df, vars_categoricas)
```

En el frame anterior se puede notar las variables categóricas, sus frecuencias, así como los valores faltanes. Es importante mencionar que la fila 0, folio, se debe ignorar ya que esta no arroja información relevante, caso contrario a las demás variables.
