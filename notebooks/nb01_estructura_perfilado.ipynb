Hagan las conversiones necesarias y expliquen cada decisión.
Se decidió realizar la conversión en las variables 'fecha_hora_creacion' y 'fecha_hora_cierre' para convertirlas a un formato de datetime64[us] que sea reconocible por Python.

```python
df['fecha_hora_creacion'] = pd.to_datetime(df['fecha_hora_creacion'] , errors='coerce')
df['fecha_hora_cierre'] = pd.to_datetime(df['fecha_hora_cierre'] , errors='coerce')     
```
